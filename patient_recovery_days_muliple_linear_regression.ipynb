```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
```

```python
# Load Data
data = {
    'age': [25, 47, 52, None, 38, 61, 29, 74, 45, 33, 58, 41, None, 67, 36],
    'bmi': [22.1, 31.4, 28.7, 24.5, None, 35.2, 21.8, 29.3, 26.4, 23.7, 32.1, 27.8, 25.6, 30.4, None],
    'blood_pressure': [120, 145, 138, 112, 155, 162, 118, 171, 134, 121, 158, 129, 141, 168, 115],
    'treatment_type': ['Surgery', 'Medication', 'Therapy', 'Medication', 'Surgery', 'Surgery', 'Therapy', 'Medication', 'Surgery', 'Therapy', 'Medication', 'Surgery', 'Therapy', 'Medication', 'Therapy'],
    'recovery_days': [45, 28, 19, 22, 52, 61, 15, 35, 48, 17, 31, 43, 18, 33, 16]
}

df = pd.DataFrame(data)
features = df.iloc[:, :-1].values
results = df.iloc[:, -1].values
```

```python
# Encode non numerical data
transformer = ColumnTransformer([('encoder', OneHotEncoder(drop='first'), [3])], remainder='passthrough')
features_encoded = transformer.fit_transform(features)
```

```python
# Split data into training and testing data
train_features, test_features, train_results, test_results = train_test_split(features_encoded, results, test_size=0.2, random_state=42)
```

```python
# Data Preprocessing
# Imputation
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
train_features_imputed = imputer.fit_transform(train_features)
test_features_imputed = imputer.transform(test_features)
```

```python
# Train Model & Predict
multi_reg = LinearRegression()
multi_reg.fit(train_features_imputed, train_results)
test_results_predicted = multi_reg.predict(test_features_imputed)
train_results_predicted = multi_reg.predict(train_features_imputed)
```

```python
# Compare actual vs predicted data
# Test data comparison
np.set_printoptions(precision=2)
print(np.concatenate((test_results.reshape(len(test_results), 1), test_results_predicted.reshape(len(test_results_predicted), 1)), axis=1))
```

```python
# Train data comparison
np.set_printoptions(precision=2)
print(np.concatenate((train_results.reshape(len(train_results), 1), train_results_predicted.reshape(len(train_results_predicted), 1)), axis=1))
```
