```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
```

```python
data = {
    'age': [25, 30, None, 40, 45, 47, 50, None, 58, 60, 63, 67, 70, None, 80],
    'blood_pressure': [112, 118, 122, 126, 131, 134, 138, 142, 146, 150, 153, 158, 161, 165, 172]
}

# Load Data into dataframe
df = pd.DataFrame(data)
age_df = df.iloc[:, :-1].values
bp_df = df.iloc[:, -1].values
```

```python
# Insert data into missing age values
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
age_df = imputer.fit_transform(age_df)
```

```python
# Split data into training and test sets
age_train, age_test, bp_train, bp_test = train_test_split(age_df, bp_df, test_size=0.2, random_state=42)
```

```python
# Scale the age values using standardization
scaler = StandardScaler()
age_train_scaled = scaler.fit_transform(age_train)
age_test_scaled = scaler.transform(age_test)
```

```python
# Fit Model
regressor = LinearRegression()
regressor.fit(age_train_scaled, bp_train)
```

```python
# Predict training and test data using model
predicted_bp_train = regressor.predict(age_train_scaled)
predicted_bp_test = regressor.predict(age_test_scaled)
```

```python
# Plot training data vs regression line graph
```

```python
plt.scatter(age_train_scaled, bp_train, color='blue')
plt.plot(age_train_scaled, predicted_bp_train, color='red')
plt.title("Age vs Blood Pressure(training data)")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.show()
```

```python
plt.scatter(age_test_scaled, bp_test, color='blue')
plt.plot(age_test_scaled, predicted_bp_test, color='red')
plt.title("Age vs Blood Pressure(test data)")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.show()
```
