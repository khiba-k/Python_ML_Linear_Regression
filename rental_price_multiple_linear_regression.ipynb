```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
```

```python
# Load Data
data = {
    'size_sqm': [45, 82, 120, None, 65, 200, 55, 95, 110, 75, None, 88, 150, 60, 170],
    'num_bedrooms': [1, 2, 3, 2, 1, 4, 1, 2, 3, 2, 2, 2, 3, 1, 4],
    'distance_to_city_km': [2.1, 5.4, 8.2, 3.7, None, 12.5, 1.8, 6.3, 9.1, 4.2, 7.8, 5.9, 11.2, 2.5, None],
    'property_type': ['Apartment', 'House', 'House', 'Apartment', 'Apartment', 'House', 'Apartment', 'House', 'House', 'Apartment', 'House', 'Apartment', 'House', 'Apartment', 'House'],
    'monthly_rent': [650, 1100, 1450, 780, 590, 2200, 700, 1250, 1380, 950, 1050, 1150, 1800, 620, 2100]
}

df = pd.DataFrame(data)

features = df.iloc[:, :-1].values
results = df.iloc[:, -1].values
```

```python
# Encode Categorical Data
ct = ColumnTransformer([('encoder', OneHotEncoder(drop='first'), [3])], remainder='passthrough')
features_encoded = ct.fit_transform(features)
```

```python
# Split data into training & test data
features_train, features_test, results_train, results_test = train_test_split(features_encoded, results, test_size=0.2, random_state=42)
```

```python
# Impute the missing feature values
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
features_train = imputer.fit_transform(features_train)
features_test = imputer.transform(features_test)
```

```python
# Train Model & Test
linear_reg = LinearRegression()
linear_reg.fit(features_train, results_train)

# Predict Results with Test Features
predicted_test = linear_reg.predict(features_test)

# Predict Results with Train Features
predicted_train = linear_reg.predict(features_train)
```

```python
# Compare Actual Results vs Predicted Results
# Test Data
np.set_printoptions(precision=2)
print(np.concatenate((results_test.reshape(len(results_test), 1), predicted_test.reshape(len(predicted_test), 1)), axis=1))
```

```python
# Train Data
np.set_printoptions(precision=2)
print(np.concatenate((results_train.reshape(len(results_train), 1), predicted_train.reshape(len(predicted_train), 1)), axis=1))
```
